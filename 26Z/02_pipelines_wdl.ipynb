# Tworzenie potoków przetwarzania danych

Na dzisiejszych zajęciach jak również na zajęciach z genomiki mieliśmy okazję zapoznać się z programami służącymi do obróbki danych pochodzących z sekwencjonowania wysokoprzepustowego. Jest to proces wieloetapowy, gdzie wykorzystywane są różne narzędzia a operacje wykonywane są na różnych rodzajach plików wejściowych generując nowego rodzaju dane. Biorąc pod uwagę mnogość etapów analizy, jak również konieczność analizy wielu plików na raz ręczne wykonywanie kolejnych etapów analizy jeden po drugim staje się niepraktyczne i nieefektywne. W odpowiedzi na te wyzwania stworzono programy do zarządzania przepływem danych, tzw. języki definicji przepływu pracy (ang. Workflow languages).

## Najpopularniejsze rodzaje języków definicji przepływu pracy
Języki definicji przepływu pracy zapewniają ustrukturyzowane ramy do opisywania i organizowania serii zadań wymaganych do analizy danych. Istnieje wiele języków definiowania przepływów pracy, używanych do pisania potoków obliczeniowych.

Common Workflow Language (lub CWL) to język służący do definiowania przepływów pracy w sposób wieloplatformowy. CWL zapewnia prosty i dobrze zdefiniowany format do automatyzacji tych analiz poprzez określenie ich etapów i połączeń za pomocą plików CWL. Ekosystem CWL powiększył się o narzędzia do wizualizacji przepływu pracy, graficzne edytory przepływu pracy, biblioteki do programowej interakcji z CWL oraz narzędzia konwertujące do i z CWL oraz innych formatów przepływu pracy.

* Oddziela definicje narzędzi i przepływy pracy, aby umożliwić ponowne wykorzystanie komponentów.
* Silna społeczność i skupienie na standardach, często postrzegane jako „klej” do łączenia narzędzi bioinformatycznych w różnych systemach.
* Zaprojektowany z myślą o szerokiej kompatybilności, umożliwiając uruchomienie tego samego przepływu pracy na różnych platformach przy minimalnych zmianach.

Tryb `server` działa jako usługa z interfejsem API HTTP, umożliwiając jednoczesne zarządzanie i monitorowanie wielu przepływów pracy. Tryb ten nadaje się do rozwiązań produkcyjnych lub dla wielu użytkowników, umożliwiając automatyczne przesyłanie, monitorowanie w czasie rzeczywistym i kolejkowanie przepływów pracy. ### Sharding – podział danych na niezależne fragmenty

W zadaniu 7_4 każda próbka z rodziny była przetwarzana przez osobną instancję zadania. Jest to najprostszy przykład **shardingu**, czyli podziału danych wejściowych na niezależne fragmenty (ang. *shards*), które mogą być przetwarzane równolegle. Taki schemat nazywany jest **scatter-gather**:

* **scatter** (rozproszenie) – dane dzielone są na fragmenty, a dla każdego z nich uruchamiana jest osobna instancja tego samego zadania,
* **gather** (zebranie) – wyniki ze wszystkich fragmentów są zbierane i, jeśli to potrzebne, scalane w jeden plik wynikowy.

Idea jest taka sama jak w przypadku partycji w Apache Spark, z którymi pracowaliśmy na wcześniejszych zajęciach. Różnica polega na tym, że Spark sam dzieli dane na partycje i rozdziela je pomiędzy executory, natomiast w języku definicji przepływu pracy to autor potoku decyduje, według jakiego klucza podzielić dane, na ile fragmentów i w jaki sposób scalić wyniki.

W analizie danych genomowych dane dzieli się najczęściej na następujących poziomach:

| Poziom podziału | Przykładowe etapy | Na co dzielimy dane | Scalanie wyników |
|---|---|---|---|
| Próbki | cały potok (zadanie 7_4) | lista plików FASTQ/BAM | osobne wyniki dla każdej próbki lub wspólne genotypowanie |
| Regiony genomu (chromosomy, interwały) | wykrywanie wariantów (HaplotypeCaller), rekalibracja jakości zasad (BQSR) | pliki BED / interval_list | `MergeVcfs`, `GatherBQSRReports` |
| Fragmenty odczytów | uliniowienie (`bwa mem`) | części pliku FASTQ | `samtools merge` |

Sharding należy odróżnić od wielowątkowości w obrębie jednego zadania (np. `bwa mem -t 4`), gdzie jeden proces korzysta z wielu rdzeni tej samej maszyny. Shardy są niezależnymi zadaniami, dlatego silnik wykonawczy może je uruchomić na różnych maszynach klastra lub w chmurze.

Sharding nie jest jednak darmowy. Każdy shard to osobne zadanie, które trzeba zaplanować, któremu trzeba udostępnić pliki wejściowe (np. cały genom referencyjny) i w którym trzeba od nowa uruchomić program (w przypadku GATK – maszynę wirtualną Javy). Do tego dochodzi etap scalania wyników. Przy małych danych lub zbyt dużej liczbie shardów ten narzut może być większy niż zysk z równoległego przetwarzania.

### Scatter i gather w WDL

<div>
<img src="https://openwdl.org/docs/patterns/scatter-gather/header.png"
width=400/>
</div>

[Wzorzec scatter-gather w dokumentacji WDL](https://openwdl.org/docs/start/patterns/scatter-gather/)

W WDL etap gather jest niejawny: jeśli zadanie zostało wywołane wewnątrz bloku `scatter`, to poza tym blokiem jego wyniki są automatycznie zbierane w tablicę. Wynik typu `File` staje się tablicą `Array[File]`, a w przypadku dwóch zagnieżdżonych bloków `scatter` – tablicą tablic `Array[Array[File]]`. Taką tablicę można przekazać jako dane wejściowe do kolejnego zadania, które scali wyniki.

Przydatne funkcje biblioteki standardowej WDL:

| Funkcja | Działanie | Przykład |
|---|---|---|
| `length(a)`, `range(n)` | długość tablicy oraz tablica indeksów `0, 1, ..., n-1` | `scatter (i in range(length(fastq_files)))` |
| `zip(a, b)` | łączy dwie tablice tej samej długości w pary (`Pair`): i-ty element z i-tym elementem | `zip(sample_names, fastq_files)` |
| `cross(a, b)` | tworzy pary ze wszystkich kombinacji elementów dwóch tablic | `cross(samples, regions)` |
| `flatten(a)` | spłaszcza tablicę tablic `Array[Array[X]]` do `Array[X]` | `flatten(ProcessRegion.result)` |
| `glob("wzorzec")` | w sekcji `output` zadania zwraca tablicę plików pasujących do wzorca | `glob("intervals/*.bed")` |
| `read_lines(plik)` | wczytuje plik tekstowy jako tablicę `Array[String]`, np. listę ścieżek do plików z interwałami | `read_lines(intervals_list)` |

Do elementów pary `Pair` odwołujemy się przez `.left` i `.right`.

Działanie tych konstrukcji zobaczymy na prostym przykładzie `dags/scatter-demo.wdl`, który zamiast programów bioinformatycznych używa polecenia `echo`:

```python
! cat dags/scatter-demo.wdl
```

### Sharding wykrywania wariantów po regionach genomu

Podział po próbkach nie przyspiesza analizy pojedynczej próbki, a w przypadku sekwencjonowania całego genomu sam etap wykrywania wariantów dla jednej próbki może trwać wiele godzin. Dlatego w praktyce dzieli się również genom referencyjny na regiony i wykrywa warianty w każdym z nich niezależnie. Jest to możliwe, ponieważ:

* dzięki indeksowi pliku BAM (plik `.bai`) program może odczytać tylko odczyty z wybranego regionu bez przeglądania całego pliku,
* HaplotypeCaller wykrywa warianty lokalnie, w tzw. aktywnych regionach, więc wynik w jednym fragmencie genomu nie zależy od odczytów z odległych fragmentów.

Potok `dags/hc-sharded.wdl` składa się z następujących etapów:

1. `BwaMem` i `SamToBam` – uliniowienie i konwersja do formatu BAM. Zadania nie są definiowane od nowa, tylko importowane z pliku `alignment.wdl` poleceniem `import "alignment.wdl" as align` i wywoływane jako `align.BwaMem` oraz `align.SamToBam` (podobnie jak moduły w potokach nf-core),

#### Na co uważać przy dzieleniu genomu na regiony

* **Granice shardów.** Nasz prosty podział tnie kontigi w dowolnych miejscach. Jeśli granica przetnie aktywny region, wariant leżący tuż przy niej może zostać pominięty lub wykryty inaczej niż przy analizie całego genomu. Dlatego w praktyce genom dzieli się w miejscach przerw w sekwencji referencyjnej (odcinków `N`), np. narzędziami GATK `ScatterIntervalsByNs` i `SplitIntervals`.
* **Wiele próbek.** Przy analizie rodziny lub kohorty shardy tworzy się zwykle dla każdej kombinacji próbka × region (zagnieżdżony `scatter` lub `cross()`). HaplotypeCaller zapisuje wtedy wyniki w formacie gVCF (`-ERC GVCF`), a wspólne genotypowanie (`GenomicsDBImport`, `GenotypeGVCFs`) również wykonywane jest osobno dla każdego regionu.
* **Nie każdy etap da się podzielić po regionach.** Na przykład oznaczanie duplikatów (`MarkDuplicates`) wymaga dostępu do wszystkich odczytów próbki (w tym par odczytów, których części mogą leżeć w różnych regionach), dlatego wykonywane jest dla całej próbki.
* **Sposób scalania zależy od rodzaju danych.** Pliki VCF z rozłącznych regionów można po prostu połączyć (`MergeVcfs`, `bcftools concat`), ale np. raporty z rekalibracji jakości zasad (BQSR) wymagają specjalnego narzędzia `GatherBQSRReports`.

W ten sam sposób zbudowany jest produkcyjny potok GATK [haplotypecaller-gvcf-gatk4.wdl](https://github.com/gatk-workflows/gatk4-germline-snps-indels/blob/master/haplotypecaller-gvcf-gatk4.wdl): lista plików z interwałami wczytywana jest funkcją `read_lines()`, HaplotypeCaller uruchamiany jest w bloku `scatter (interval_file in scattered_calling_intervals)`, a wyniki scalane są w zadaniu `MergeGVCFs`. Więcej potoków produkcyjnych napisanych w WDL można znaleźć w repozytorium [WARP](https://github.com/broadinstitute/warp) (Broad Institute).
